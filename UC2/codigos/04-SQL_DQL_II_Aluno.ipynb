# Integração Python + MySQL

A integração entre Python e MySQL é feita principalmente através do driver oficial `mysql-connector-python`, que permite executar qualquer comando SQL diretamente do seu código.

## 1. Instalação do Driver
Python para se conectar a um banco de dados MySQL, divididas por categoria de uso:


SQL Manual foca em controle absoluto e velocidade. É a escolha ideal para scripts diretos, migrações de dados, relatórios pesados ou quando você já tem domínio da linguagem SQL e quer extrair a máxima performance do banco sem intermediários.  

ORM (Classes Python) foca em produtividade e segurança. É o padrão para o desenvolvimento de APIs e sistemas web, pois transforma tabelas em classes Python, previne ataques de SQL Injection automaticamente e facilita a manutenção do código a longo prazo.  

*Tabela Comparativa*


|Critério	|SQL Manual (PyMySQL, mysqlclient)	|ORM (SQLAlchemy, Peewee)|
|-----|-----|-----|
|Curva de Aprendizado	|Baixa (se você já conhece SQL).	|Média (precisa aprender a sintaxe do framework).|
|Performance	|Máxima, sem intermediários.	|Ligeiramente menor devido à tradução do código.|
|Segurança (SQL Injection)	|Requer cuidado manual do desenvolvedor.	|Automática na maioria das operações.|
|Manutenção do Código	|Difícil em projetos grandes com muitas tabelas.	|Fácil, centralizada em classes Python.|
|Troca de Banco de Dados	|Difícil (exige reescrever consultas específicas).	|Fácil (basta alterar a string de conexão).|
|Indicado Para	|Scripts rápidos, automações e relatórios pesados.	|APIs, aplicações Web (FastAPI, Flask) e sistemas complexos.|

## 2. Criando a Conexão e o Cursor

O primeiro passo no código é estabelecer a conexão utilizando as credenciais do seu servidor MySQL. Em seguida, criamos um objeto cursor, que é o responsável por executar as instruções SQL.

#### Sintaxe:

        import mysql.connector  

1. Estabelece a conexão  :

        conexao = mysql.connector.connect(  
            host="localhost",       # Endereço do servidor (ex: localhost ou 127.0.0.1)  
            user="root",            # Seu usuário do MySQL  
            password="sua_senha",   # Sua senha do MySQL  
            database="seu_banco"    # Nome do banco de dados que deseja acessar  
        )  

2. Cria o cursor para executar comandos  
   
        cursor = conexao.cursor()

Usar o Pandas junto com o SQLAlchemy é a forma mais prática de trabalhar com dados no Python, pois o Pandas lê o resultado da consulta e já o transforma diretamente em uma tabela (DataFrame) pronta para análise.

Exemplo completo e direto utilizando essa combinação:

```python
%pip install sqlalchemy
%pip install pymysql
```

```python
import pandas as pd
from sqlalchemy import create_engine # biblioteca para criar a engine de conexão com o banco de dados


# 1. Variáveis com os dados de conexão do seu banco
USER = "aluno_consulta"
PASSWORD = "SenhaSeguraAluno2026!"
HOST = "mysql-13e2d8cd-educacao-09e9.b.aivencloud.com"
PORT = 21972
DB = "analise_vendas"

# 2. Monta a string de conexão usando f-string de forma limpa e organizada
url_conexao = f"mysql+pymysql://{USER}:{PASSWORD}@{HOST}:{PORT}/{DB}"

# 3. Cria o motor (engine) do SQLAlchemy injetando a URL e o parâmetro SSL
engine = create_engine( #  cria a engine de conexão com o banco de dados
    url_conexao, # URL de conexão
    connect_args={"ssl": {"ssl_mode": "REQUIRED"}} # necessário para conexões seguras com o Aiven
)

# 4. Define a consulta SQL
query = "SELECT nome, cidade, segmento FROM clientes LIMIT 10;"

# 5. O Pandas executa a consulta e armazena os dados em um DataFrame
df = pd.read_sql(query, con=engine) # Executa a consulta SQL e armazena os resultados em um DataFrame

# 6. Exibe o DataFrame formatado
print("DataFrame dos Clientes:")
print(df)
```

Para filtrar os dados diretamente no Pandas (sem precisar fazer uma nova consulta ao banco de dados), você pode usar os recursos nativos do DataFrame. Isso economiza processamento no servidor do MySQL, já que você manipula os dados que já estão na memória do Python.  

Vamos ver as 3 formas mais comuns de filtrar os dados da tabela gerada:

```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Configuração da conexão (suas variáveis)
USER = "aluno_consulta"
PASSWORD = "SenhaSeguraAluno2026!"
HOST = "mysql-13e2d8cd-educacao-09e9.b.aivencloud.com"
PORT = 21972
DB = "analise_vendas"

url_conexao = f"mysql+pymysql://{USER}:{PASSWORD}@{HOST}:{PORT}/{DB}"
engine = create_engine(url_conexao, connect_args={"ssl": {"ssl_mode": "REQUIRED"}})

# 2. Puxa os dados do banco (trazendo mais linhas para podermos filtrar)
query = "SELECT nome, cidade, segmento FROM clientes LIMIT 100;"
df = pd.read_sql(query, con=engine)

print("--- DataFrame Original (Primeiras Linhas) ---")
print(df.head())


# =====================================================================
# ABORDAGEM 1: Filtro por texto exato (Mapeamento Direto)
# =====================================================================
# Filtra apenas clientes onde a coluna 'segmento' é exatamente igual a 'Corporativo'
df_corporativo = df[df['segmento'] == 'Corporativo']

print("\n--- Clientes do Segmento Corporativo ---")
print(df_corporativo)


# =====================================================================
# ABORDAGEM 2: Filtro por texto parcial (Contém / Like do SQL)
# =====================================================================
# Filtra cidades que contenham a palavra 'São' (ex: São Paulo, São José)
# O 'na=False' evita erros se houver valores nulos na coluna
df_sao = df[df['cidade'].str.contains('São', na=False)]

print("\n--- Clientes em cidades que começam/contêm 'São' ---")
print(df_sao)


# =====================================================================
# ABORDAGEM 3: Múltiplos Filtros (E / OU)
# =====================================================================
# Filtra clientes do segmento 'Corporativo' E que moram no 'Rio de Janeiro'
# Use '&' para E (todas condições verdadeiras) e '|' para OU (pelo menos uma verdadeira)
filtro_composto = df[(df['segmento'] == 'Corporativo') & (df['cidade'] == 'Rio de Janeiro')]

print("\n--- Clientes Corporativos no Rio de Janeiro ---")
print(filtro_composto)
```


## 3. Operações Básicas (CRUD)


🟢 Create (Inserir Dados)  

Para enviar dados ao banco, use o método execute().  
É fundamental utilizar placeholders (%s) para evitar ataques de SQL Injection.  
Sempre chame o conexao.commit() para salvar as alterações.

```python
sql = "INSERT INTO usuarios (nome, email) VALUES (%s, %s)"
valores = ("Ana Silva", "ana.silva@email.com")

cursor.execute(sql, valores)
conexao.commit()  # Confirma a alteração no banco de dados

print(f"Registro inserido com sucesso! ID: {cursor.lastrowid}")
cursor.close()
conexao.close()
```


🔵 Read (Buscar Dados)

Após executar um comando SELECT, você deve utilizar o método fetchall() ou fetchone() para coletar os dados retornados pelo servidor.

```python
cursor.execute("SELECT id, nome, email FROM usuarios") # Executa a consulta SQL
resultados = cursor.fetchall()  # Retorna uma lista de tuplas

for linha in resultados: # Itera sobre cada linha do resultado
    print(f"ID: {linha[0]} | Nome: {linha[1]} | Email: {linha[2]}")
```


🟡 Update (Atualizar Dados)

Modifique registros existentes também fazendo o uso de parâmetros e confirmando a ação com o commit

```python
sql = "UPDATE usuarios SET email = %s WHERE id = %s"
valores = ("ana.novoemail@email.com", 1)

cursor.execute(sql, valores)
conexao.commit()

print(f"{cursor.rowcount} registro(s) atualizado(s).")
cursor.close()
```


🔴 Delete (Apagar Dados)

Exclua registros seguindo a mesma estrutura segura.

```python
sql = "DELETE FROM usuarios WHERE id = %s"
valor = (1,)  # Nota: é necessário passar como tupla, por isso a vírgula

cursor.execute(sql, valor)
conexao.commit()

print(f"{cursor.rowcount} registro(s) excluído(s).")
```


## 4. Fechando as Conexões

Após concluir as operações, é uma boa prática fechar o cursor e a conexão para liberar os recursos do servidor.

```python
cursor.close() # Fecha o cursor
conexao.close() # Fecha a conexão com o banco de dados
```
